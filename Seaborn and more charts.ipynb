```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# **EDA**

```python
df = spark.read.csv(
    "/Volumes/workspace/default/brightlearn-retail/1772721563345_retail_sales_dataset(1).csv",
    header=True,
    inferSchema=True
)

df.show()
```

```python
spark_df = spark.sql("select * from default.brightlearn_retail")
df = spark_df.toPandas()

#Convert date column
df["date"] = pd.to_datetime(df["date"])

df.head()
```

```python
df.info()
```

# **Graphs**

```python
revenue_by_product = df.groupby("product_category")["total_amount"].sum()
```

# **Barplot**

```python
revenue_by_product.plot(kind="bar")
```

```python
sns.set_style("whitegrid")
```

```python
sns.barplot(x=revenue_by_product.index, y=revenue_by_product.values)
plt.title("Revenue by product")
plt.xticks(rotation = 45)
plt.show()
```

# **Line plot**

```python
revenue_over_time = df.groupby("date")["total_amount"].sum().reset_index()

sns.lineplot(data = revenue_over_time, x="date", y = "total_amount")

plt.show()
```

# **Multi-line plot**

```python
sns.lineplot(data=df, x="date", y="total_amount", hue="product_category")
```

```python
pivot = df.groupby(["date","product_category"])["total_amount"].sum().unstack()
pivot.plot()
plt.show()
```

# **Pie Chart**
Pie chart are risky. only use the for proportions

- values-size of slices
- labels-names
- autopct-shows%
- startangle-rotates chart

```python
plt.figure(figsize=(6,6))

plt.pie( revenue_by_product.values,
        labels = revenue_by_product.index,
        autopct='%1.1f%%',
        startangle = 90
        )

plt.title("Revenue contribution by Product")
plt.show()
```

# **Primary & Secondary axes : Dual- Axis chart**

```python
revenue = df.groupby("date")["total_amount"].sum()
transactions = df.groupby("date")["quantity"].sum()
```

```python
# Create figure
fig, ax1 = plt.subplots(figsize=(10,5))

# -----------------------------
# First axis (Revenue)
# -----------------------------
ax1.plot(revenue.index, revenue.values, color="green")
ax1.set_xlabel("Date")
ax1.set_ylabel("Revenue")
ax1.tick_params(axis='y', labelcolor="green")

# -----------------------------
# Second axis (Transactions)
# -----------------------------
ax2 = ax1.twinx()

ax2.bar(transactions.index, transactions.values, alpha=0.3)
ax2.set_ylabel("Number of Transactions")

# Title
plt.title("Revenue vs Transactions Over Time")

plt.show()
```
