```python
df = spark.read.csv(
    "/Volumes/workspace/default/brightlearn-retail/1772721563345_retail_sales_dataset(1).csv",
    header=True,
    inferSchema=True
)

df = df.toDF(*[c.lower().replace(" ", "_") for c in df.columns])

df.write.mode("overwrite").saveAsTable("default.brightlearn_retail")
```

**pandas data frame**

```python
spark_df = spark.sql("select * from default.brightlearn_retail")
df = spark_df.toPandas()
df.head()
```

**revenue by product**


```python
revenue_by_product = df.groupby("product_category") ["total_amount"].sum()
print("revenue_by_product: ")
print(revenue_by_product)
```

# **Matplotlib with Pandas**

**First Visualization**

```python
revenue_by_product.plot(kind="bar")
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.show()
```

**Time Based Analysis (Line Chart)**

```python
df["date"] = pd.to_datetime(df["date"])
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["weekday"] = df["date"].dt.weekday
df["hour"] = df["date"].dt.hour
df["minute"] = df["date"].dt.minute
df["second"] = df["date"].dt.second
df.info()
```

```python
revenue_over_time = df.groupby("date") ["total_amount"].sum()
print("revenue_over_time: ")
print(revenue_over_time)
revenue_over_time.plot(kind="line")
plt.title("Revenue over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```

```python
revenue_overall = df.groupby("year") ["total_amount"].sum()
print("revenue_overall: ")
print(revenue_overall)
revenue_overall.plot(kind="bar")
plt.title("Revenue by Year")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show() 

```

# **using Matplotlib without pandas to plot the graph**

we dont you kind but e use the name/type of a graph as function

```python
plt.bar(revenue_by_product.index, revenue_by_product.values)
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=90)
```

**we unstack()- this converts the product categories into columns.
Each colomn is a separate line**

```python
pivot = df.groupby(["date", "product_category"])["total_amount"].sum().unstack()
pivot.plot()
plt.show()
```

```python
pivot = df.groupby(["date", "product_category"])["total_amount"].sum().unstack("product_category")
pivot.plot(kind="bar", stacked=True)
plt.title("Revenue by Product and Date")
plt.xlabel("Date")
plt.ylabel("Revenue")
```
